# AI · Day 1 Exercise
**Task:** Write a simple Python rule-based program that identifies whether a message contains an AI-related keyword.

**Dataset:** Python Basics / Synthetic Messages

````python
import pandas as pd

# Synthetic Messages dataset (Python Basics / Synthetic Messages)
messages = [
    "I just booked a Swiggy delivery for lunch",
    "This chatbot answered my query instantly",
    "The new machine learning model beat our old system",
    "Can you send me the meeting notes from yesterday?",
    "Our neural network needs more training data",
    "I watched a cricket match last night, India won!",
    "The AI assistant scheduled my flight automatically",
    "Please pick up milk on your way home",
    "This algorithm predicts customer churn with 92% accuracy",
    "Happy birthday! Hope you have a great day",
    "We are deploying a deep learning model for image recognition",
    "The weather looks great for a walk today",
    "Our robot vacuum mapped the whole house by itself",
    "Let's catch up over coffee this weekend",
    "The recommendation engine uses AI to suggest movies",
]

print(f"Total messages: {len(messages)}")
````

````python
# Rule-based keyword detector
AI_KEYWORDS = [
    "ai", "artificial intelligence", "machine learning", "ml",
    "neural network", "deep learning", "chatbot", "algorithm",
    "robot", "automat",  # matches automatic / automation
]

def contains_ai_keyword(message: str) -> bool:
    text = message.lower()
    return any(keyword in text for keyword in AI_KEYWORDS)

df = pd.DataFrame({"message": messages})
df["is_ai_related"] = df["message"].apply(contains_ai_keyword)

df
````

````python
print(f"AI-related messages: {df['is_ai_related'].sum()} / {len(df)}")
````

## Interpretation

The rule-based detector correctly flagged 8 of 15 messages as AI-related just by checking for keyword substrings like "chatbot," "algorithm," and "neural network" — no machine learning required for this simple case. This approach is fast and fully explainable, but it's brittle: it would miss a message like "the system learned from our data on its own" (AI-related, but no keyword match) and could misfire on a message that happens to contain a keyword out of context. This is exactly the gap that real machine learning models are built to close — they learn patterns instead of relying on exact keyword matches.

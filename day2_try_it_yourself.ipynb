# Try It Yourself: Spot the Invisible AI

**Exercise:** Write a simple Python rule-based program that scans a set of everyday app notifications and flags which ones show AI quietly making a decision *for* you (like the Swiggy, Hotstar, and Face ID examples from Day 2's post).

**Dataset:** Synthetic Messages (app notifications)

**Hint:** Build a small list of "invisible AI" signal phrases — "recommended for you", "predicted", "detected", "auto-selected", "personalized" — and check if each message contains one.

```python
# Synthetic Messages: everyday app notifications
messages = [
    "Your order has been auto-selected based on your past picks",
    "Reminder: your electricity bill is due on the 15th",
    "We predicted you'd love this show — new episode is out",
    "Meeting rescheduled to 4 PM tomorrow",
    "This route was recommended for you based on current traffic",
    "Happy Birthday from all of us at the store!",
    "Unusual login detected on your account just now",
    "Your feed has been personalized using your recent activity",
    "Please carry a valid ID for tomorrow's exam",
    "5 items are frequently bought together with your last order"
]

# Small list of "invisible AI" signal phrases
ai_signals = ["recommended for you", "predicted", "detected", "auto-selected", "personalized"]

def has_ai_signal(message):
    text = message.lower()
    return any(signal in text for signal in ai_signals)

flagged = [(msg, has_ai_signal(msg)) for msg in messages]

for msg, flag in flagged:
    print(f"[{'AI' if flag else '--'}] {msg}")

count_flagged = sum(flag for _, flag in flagged)
print()
print(f"Messages flagged as AI-driven: {count_flagged} / {len(messages)}")
```

## Interpretation

Half of the 10 messages (5/10) were flagged as AI-driven, showing how often a system is quietly recommending, predicting, or detecting something on your behalf without ever saying "AI" out loud. A simple keyword check like this is a useful first filter, but it's rule-based, not intelligent — it would miss an AI-driven message phrased differently (e.g. "picked just for you") and could wrongly flag a human-written message that happens to use one of the words. This is exactly why real systems use more than plain keyword matching, but keyword matching is still a great first step to build intuition.
